```python
import glob
import pandas as pd
import os
import numpy as np
import re
import matplotlib.pyplot as plt

#year = 2012

path = '/n/netscratch/cga/Lab/dkakkar_1/ny_tweets_ida/' # use your path
files = glob.glob(path + "/*.gz")
# Read in hourly files

#print (files)
parts = []
df = pd.DataFrame()

for file in files:
    print(file)

    try:
           temp = pd.read_csv(file, sep='\t',dtype='unicode',index_col=None, low_memory='true',compression='gzip')
    except Exception as e:
            print(e,"can't read!")
            #l_ni.append(filename)
            continue
    
    #print(temp['latitude'])
    
    #temp.columns=['message_id','tweet_date','tweet_text','tags','tweet_lang','source','place','retweets','tweet_favorite', 'photo_url','quoted_status_id','user_id','user_name','user_location','followers','friends','user_favorites','status','user_lang','latitude','longitude','data_source','GPS','spatialerror'] 
    cols = ['latitude', 'longitude']
    temp[cols] = temp[cols].apply(pd.to_numeric, errors='coerce')
    #print (temp.head())
    
    for i in range(len(temp['message_id'])):
        if temp['longitude'][i]=="True":
            temp['longitude'][i] = np.nan
            
    lat_min, lat_max = 38.9285, 41.3572
    lon_min, lon_max = -75.5598, -73.9024
       # make sure longitude is numeric (if it's string in your df)
    temp['longitude'] = temp['longitude'].astype(float)

    # filter into a new DataFrame
    ny_temp = temp[
    (temp['latitude'].between(lat_min, lat_max)) &
    (temp['longitude'].between(lon_min, lon_max))
    ]

# check result
    #print(ny_temp.shape)   # how many rows
    #print(ny_temp.head())  # preview
    
     
    #print(df.head())
    
    pattern = r'(?i)\b(?:ida|flood(?:ing)?|climate(?:\s+change)?|fema|disaster|crisis|hurricane|typhoon|extreme\s+weather|rescue|mitigation|adaptation|rain|tropical|tornado|ef-?3|storm|assistance|aid|recovery|resilience|executive\s+order)\b'

    ny_temp_Ida = ny_temp[ny_temp['text'].str.contains(pattern, na=False, regex=True)]

    
    pattern_1=r"#?(?:NJ)"
    
    ny_temp_Ida_NJ = ny_temp_Ida[
    ny_temp_Ida['place'].str.contains(pattern_1, case=False, na=False, regex=True)]
    
    #parts.append(ny_temp)
    parts.append(ny_temp_Ida_NJ)  # <-- collect, don't concat each loop

# final dataframe with all files appended
df = pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()
print(df.shape)
df.to_csv("/n/netscratch/cga/Lab/dkakkar_1/ny_tweets_ida/ny_temp_Ida_2_NJ.csv", index=False)
      
```
